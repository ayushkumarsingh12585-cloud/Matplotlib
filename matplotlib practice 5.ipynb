```python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr"]
sales = [200, 250, 230, 300]

plt.figure(figsize=(10, 7))

plt.subplot(2, 2, 1)
plt.plot(months, sales, marker="o", color="blue")
plt.title("Line Chart")

plt.subplot(2, 2, 2)
plt.bar(months, sales, color="green")
plt.title("Bar Chart")

plt.subplot(2, 2, 3)
plt.scatter(months, sales, color="red")
plt.title("Scatter Plot")

plt.subplot(2, 2, 4)
plt.pie(sales, labels=months, autopct="%1.1f%%")
plt.title("Pie Chart")

plt.suptitle("Monthly Sales Dashboard", fontsize=16)
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr"]
sales = [200, 250, 230, 300]

plt.figure(figsize=(12, 8))

plt.subplot(2, 3, 1)
plt.plot(months, sales, color="blue", marker="o")
plt.title("Line Chart")

plt.subplot(2, 3, 2)
plt.bar(months, sales, color="green")
plt.title("Bar Chart")

plt.subplot(2, 3, 3)
plt.scatter(months, sales, color="red")
plt.title("Scatter Plot")

plt.subplot(2, 3, 4)
plt.pie(sales, labels=months, autopct="%1.1f%%")
plt.title("Pie Chart")

plt.subplot(2, 3, 5)
plt.plot(months, sales, color="orange", linestyle="--")
plt.title("Dashed Line Chart")

plt.subplot(2, 3, 6)
plt.barh(months, sales, color="purple")
plt.title("Horizontal Bar Chart")

plt.suptitle("Monthly Sales Dashboard", fontsize=16)
plt.tight_layout()
plt.show()
```
