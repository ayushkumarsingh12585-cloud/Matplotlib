```python
import matplotlib.pyplot as plt

product = ["A", "B", "C", "D"]
sales2025 = [100, 120, 150, 130]
sales2026 = [130, 140, 170, 160]

plt.bar(product, sales2025, label="2025")
plt.bar(product, sales2026, bottom=sales2025, label="2026")

plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt

# line graph
plt.subplot(1, 3, 1)
plt.plot([1, 2, 3, 4], [10, 20, 30, 40])
plt.title("Sales")

# bar graph
plt.subplot(1, 3, 2)
plt.bar(["A", "B", "C"], [70, 80, 90])
plt.title("Marks")

# pie chart
plt.subplot(1, 3, 3)
plt.pie([40, 30, 30], labels=["Food", "Travel", "Other"])
plt.title("Expenses")

plt.show()
```
