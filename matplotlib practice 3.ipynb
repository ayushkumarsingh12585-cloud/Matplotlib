```python
import matplotlib.pyplot as plt

month = ["Jan", "Feb", "Mar", "Apr"]

a = [100, 120, 150, 170]
b = [80, 110, 130, 160]

plt.plot(month, a, label="Product A")
plt.plot(month, b, label="Product B")

plt.legend()
plt.show()
```
