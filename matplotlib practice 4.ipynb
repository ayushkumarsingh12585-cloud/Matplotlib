```python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr"]
sales = [200, 250, 230, 300]

plt.figure(figsize=(12, 8))

plt.subplot(2, 3, 1)
plt.plot(months, sales, color="blue", marker="o")
plt.title("Line Chart")

plt.subplot(2, 3, 2)
plt.bar(months, sales, color="green")
plt.title("Bar Chart")

plt.subplot(2, 3, 3)
plt.scatter(months, sales, color="red")
plt.title("Scatter Plot")

plt.subplot(2, 3, 4)
plt.pie(sales, labels=months, autopct="%1.1f%%")
plt.title("Pie Chart")

plt.subplot(2, 3, 5)
plt.plot(months, sales, color="orange", linestyle="--")
plt.title("Dashed Line Chart")

plt.subplot(2, 3, 6)
plt.barh(months, sales, color="purple")
plt.title("Horizontal Bar Chart")

plt.suptitle("Monthly Sales Dashboard", fontsize=16)
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr"]
sales = [200, 250, 230, 300]

plt.figure(figsize=(14, 10))

plt.subplot(3, 3, 1)
plt.plot(months, sales, color="blue", marker="o")
plt.title("Line Chart")

plt.subplot(3, 3, 2)
plt.bar(months, sales, color="green")
plt.title("Bar Chart")

plt.subplot(3, 3, 3)
plt.scatter(months, sales, color="red")
plt.title("Scatter Plot")

plt.subplot(3, 3, 4)
plt.pie(sales, labels=months, autopct="%1.1f%%")
plt.title("Pie Chart")

plt.subplot(3, 3, 5)
plt.plot(months, sales, color="orange", linestyle="--")
plt.title("Dashed Line")

plt.subplot(3, 3, 6)
plt.barh(months, sales, color="purple")
plt.title("Horizontal Bar")

plt.subplot(3, 3, 7)
plt.hist(sales, color="brown", bins=4)
plt.title("Histogram")

plt.subplot(3, 3, 8)
plt.plot(months, sales, color="red", marker="*", linewidth=3)
plt.title("Line with Stars")

plt.suptitle("Monthly Sales Dashboard", fontsize=18)
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr"]
sales = [200, 250, 230, 300]

plt.figure(figsize=(8, 5))
plt.plot(months, sales, color="blue", linestyle="--", linewidth=2, marker="o", markersize=8, alpha=0.8, label="Sales")
plt.title("Monthly Sales", fontsize=16)
plt.xlabel("Month", fontsize=12)
plt.ylabel("Sales", fontsize=12)
plt.grid()
plt.legend()
plt.show()
```
